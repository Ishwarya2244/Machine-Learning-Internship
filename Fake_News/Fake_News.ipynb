```python
import pandas as pd
df=pd.read_csv('/WELFake_Dataset.csv',on_bad_lines='skip',engine='python')
```

# **prepare data**

```python
df=df.drop(columns=['Unnamed: 0'])
```

```python
df=df.fillna('0')
```

```python
df['Content']=df['title']+" "+df['text']
```

```python
X=df['Content']
Y=df['label']
```

```python
df.head()
```

# **Text to Number convert**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(stop_words='english',max_df=0.7)
```

```python
X_tfidf=tfidf.fit_transform(X)
```

# **Train and Test split**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X_tfidf,Y,test_size=0.2,random_state=42)
```

**Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression(max_iter=1000) # max_iter= give model enough time to learn
```

# **Train the model**

```python
model.fit(X_train,Y_train)
```

# **model Evaluation**

# **importing Evaluation tools**

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

# **making predications**

```python
Y_pred=model.predict(X_test) #after trained the model we must analyze how well it learned . asking the trained model to make predictions
```

```python
accuracy=accuracy_score(Y_test,Y_pred)
print("Accuracy:",accuracy)
```

## **Classification Report**

```python
print(classification_report(Y_test,Y_pred))
```

# **confusion matrix**

```python
cm=confusion_matrix(Y_test,Y_pred)
print(cm)
#TN = 8
#Real news correctly predicted as real
#FP = 16
#Real news wrongly predicted as fake
 #FN = 0
#Fake news wrongly predicted as real → NONE (very good)
#TP = 30
 #Fake news correctly predicted as fake
 #[[TN  FP]
# [FN  TP]]
```
